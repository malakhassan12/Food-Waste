## **Show ouliers**

```python
outlier_rows = []

for col in numeric_cols:
    series = egypt_food_waste[col].dropna()

    if series.empty:
        continue

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    mask = (egypt_food_waste[col] < lower_bound) | (
        egypt_food_waste[col] > upper_bound
    )

    outlier_rows.append({
        "column": col,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": int(mask.sum()),
        "outlier_percentage": round(mask.mean() * 100, 2)
    })

outlier_report = pd.DataFrame(outlier_rows)
display(outlier_report)
```

```python
q = egypt_food_waste["loss_quantity"].astype("string").str.strip()

is_per_ha  = q.str.contains("/ha", na=False)        # "0.61ton/ha"
is_million = q.str.contains(r"\bM\b", na=False)     # "4.4 M tonnes"
num = pd.to_numeric(q.str.extract(r"([\d.]+)")[0], errors="coerce")

egypt_food_waste["loss_tonnes"] = np.where(
    is_per_ha, np.nan,
    np.where(is_million, num * 1_000_000, num)
)
egypt_food_waste["loss_tonnes_per_ha"] = np.where(is_per_ha, num, np.nan)

egypt_food_waste = egypt_food_waste.drop(columns="loss_quantity")
```

# **Step 2: Finalize the dataset (flags + missing values + validation + save)**

### **1. Fix `cpc_code` (restore leading zeros)**

```python
# cpc_code was read as float, so "0111" became "111.0". All CPC codes here start with "0".
# This cell is safe to run twice (idempotent).
s = egypt_food_waste["cpc_code"].astype("string").str.strip().str.replace(r"\.0$", "", regex=True)
egypt_food_waste["cpc_code"] = s.where(s.str.startswith("0"), "0" + s)

print(egypt_food_waste["cpc_code"].unique())
```

### **2\. Detect Potential Duplicates, Flag Suspected Outliers, and Handle Zero Loss Values Without Deleting Suspicious Records**

```python
df = egypt_food_waste


key = ["commodity", "year", "food_supply_stage", "loss_percentage"]
national_keys = set(map(tuple, df.loc[df["geo_level"].eq("National"), key].values))
df["is_new_valley_copy"] = (
    df["governorate"].eq("New Valley")
    & df[key].apply(tuple, axis=1).isin(national_keys)
)

whole = df["food_supply_stage"].eq("Whole supply chain")
med   = df.loc[whole].groupby("commodity")["loss_percentage"].transform("median")
ratio = (df["loss_percentage"] / med).reindex(df.index)
df["is_suspect_value"] = whole & med.reindex(df.index).gt(0) & ((ratio < 0.2) | (ratio > 5))


df["is_zero_loss"] = df["loss_percentage"].eq(0)
df["loss_percentage_clean"] = df["loss_percentage"].mask(df["is_zero_loss"])

egypt_food_waste = df
print(df[["is_suspect_year", "is_lit_outlier", "is_new_valley_copy",
          "is_suspect_value", "is_zero_loss"]].sum())
display(df.loc[df["is_suspect_value"] | df["is_new_valley_copy"],
               ["commodity", "year", "food_supply_stage", "governorate", "loss_percentage"]])
```

### **3. Handle missing values**
| Column | Why missing | Action |
|---|---|---|
| `treatment` | only recorded for wheat harvest/storage | fill with label **"Not specified"** |
| `governorate` | row is national or regional | fill with label **"Not applicable"** |
| `notes` | no note in source | fill with label **"No note"** |
| `loss_tonnes` | source did not report quantity | **keep NaN** + indicator column |
| `loss_tonnes_per_ha` | only wheat harvest rows | **keep NaN** + indicator column |
| `loss_percentage_clean` | zero values turned into NaN | **keep NaN** |

Numeric columns are NOT imputed (mean/median would invent loss quantities that no source reported).

```python
df = egypt_food_waste

df["has_loss_tonnes"]        = df["loss_tonnes"].notna()
df["has_loss_tonnes_per_ha"] = df["loss_tonnes_per_ha"].notna()

df["treatment"]   = df["treatment"].fillna("Not specified")
df["governorate"] = df["governorate"].fillna("Not applicable")
df["notes"]       = df["notes"].fillna("No note")

df["is_reliable"] = ~(df["is_suspect_year"] | df["is_new_valley_copy"]
                      | df["is_suspect_value"] | df["is_zero_loss"])

egypt_food_waste = df
print("Reliable rows:", int(df["is_reliable"].sum()), "of", len(df))
display(df.isna().sum()[lambda s: s > 0])
```

### **4. Types, column order, validation**

```python
df = egypt_food_waste

cat_cols = ["commodity", "food_supply_stage", "treatment", "geo_level", "governorate",
            "measurement_scope", "method_data_collection", "reference"]
for c in cat_cols:
    df[c] = df[c].astype("category")

final_cols = [
    # what / when / where
    "cpc_code", "commodity", "year", "geo_level", "governorate",
    "food_supply_stage", "measurement_scope", "treatment",
    # values
    "loss_percentage", "loss_percentage_clean", "loss_percentage_original",
    "loss_tonnes", "has_loss_tonnes", "loss_tonnes_per_ha", "has_loss_tonnes_per_ha",
    # quality flags
    "is_reliable", "is_suspect_year", "is_lit_outlier", "is_new_valley_copy",
    "is_suspect_value", "is_zero_loss",
    # source
    "method_data_collection", "reference", "notes",
]
df = df[final_cols].drop_duplicates().reset_index(drop=True)

# ---- validation (will raise an error if something is wrong) ----
assert df.duplicated().sum() == 0,                       "duplicate rows found"
assert df["loss_percentage"].between(0, 100).all(),      "loss_percentage out of 0-100"
assert df["cpc_code"].str.match(r"^0\d{3,4}(\.\d{2})?$").all(), "bad cpc_code format"
assert df["year"].between(2000, 2026).all(),             "year out of range"
key_cols = ["cpc_code", "commodity", "year", "loss_percentage", "food_supply_stage",
            "measurement_scope", "geo_level", "method_data_collection", "reference"]
assert df[key_cols].notna().all().all(),                 "missing values in key columns"
print("All checks passed.", df.shape)

egypt_food_waste_final = df
display(df.head())
```

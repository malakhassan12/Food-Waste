### **2. Clean text, unify spelling, drop duplicated data**
> الداتا الخام فيها 180 صف مكرر فقط. بنشيل التكرار على **كل الأعمدة الأصلية** قبل ما نحذف أو نغيّر أي عمود، وإلا صفوف من دول مختلفة هتبان مكررة وتتمسح بالغلط.

```python
df = world_food_loss

MISSING = {"", "na", "n/a", "null", "none", "unknown", "missing", "-", "nan"}
for c in df.select_dtypes(include=["object", "string"]).columns:
    s = df[c].astype("string").str.strip().str.replace(r"\s+", " ", regex=True)
    df[c] = s.mask(s.str.lower().isin(MISSING))

# HTML tags inside some commodity names, e.g. "Chillies (<i>Capsicum</i> spp.)"
df["commodity"] = (df["commodity"].str.replace(r"<[^>]+>", "", regex=True)
                                  .str.replace(r"\s+", " ", regex=True).str.strip())

def unify_case(s):
    """'Major Season' / 'Major season' -> the most frequent spelling."""
    s = s.astype("string")
    key = s.str.casefold()
    canon = (pd.DataFrame({"k": key, "v": s}).dropna()
               .groupby("k")["v"].agg(lambda x: x.value_counts().index[0]).to_dict())
    return key.map(canon).astype("string")

for c in ["country", "commodity", "region", "treatment", "cause_of_loss", "activity",
          "food_supply_stage", "method_data_collection", "reference"]:
    df[c] = unify_case(df[c])

before = len(df)
df = df.drop_duplicates().reset_index(drop=True)
print(f"exact duplicates removed: {before - len(df)}  ->  {len(df)} rows")

# one name per country code and one name per CPC code (UK / Taiwan variants, 'n.e.' vs 'n.e.c.')
most_common = lambda x: x.value_counts().index[0]
n1, n2 = df["country"].nunique(), df["commodity"].nunique()
df["country"]   = df.groupby("m49_code")["country"].transform(most_common)
df["commodity"] = df.groupby("cpc_code")["commodity"].transform(most_common)
print("country names:", n1, "->", df["country"].nunique())
print("commodity names:", n2, "->", df["commodity"].nunique())

world_food_loss = df
```

### **4. Separate loss_quantity to 2 columns since the Unit (tonnes / tonnes per ha) and mark rates**

```python
df = world_food_loss

df["loss_quantity_text"] = df["loss_quantity"]
q = df["loss_quantity"].astype("string").str.lower().str.strip()
q = q.str.replace(r"(?<=\d),(?=\d)", "", regex=True)                  # 5,799.33 -> 5799.33
q = q.str.replace(r"(?<=\d)\s(?=\d{3}(?!\d))", "", regex=True)        # "36 275" -> 36275

num_all = pd.to_numeric(q, errors="coerce")                             # plain numbers (1e+05 included)
num_txt = pd.to_numeric(q.str.extract(r"(\d+(?:\.\d+)?)")[0], errors="coerce")
num = num_all.fillna(num_txt)

per_ha   = q.str.contains(r"/\s*ha\b|per\s*ha\b|ha-1|t/ha", na=False)
is_rate  = (q.str.contains(r"/\s*(?:mt|quintal|q\b|capita|household|acre|farm)|per\s*(?:quintal|acre|capita|household|farm)"
                           r"|\bhead\b|kfarm|kga", na=False)) & ~per_ha
million  = q.str.contains(r"million|milion|\bm\s+(?:ton|tonne|kg)", na=False)
thousand = q.str.contains(r"thousand", na=False)
kilo     = q.str.contains(r"\bkgs?\b|kilogram", na=False)
quintal  = q.str.contains(r"quintal", na=False)
unit_known = per_ha | is_rate | million | thousand | kilo | quintal | q.str.contains(r"ton|tonne|\bt\b|\bmt\b|metric", na=False)

scale  = np.where(million, 1_000_000, np.where(thousand, 1_000, 1))
factor = np.where(kilo, 0.001, np.where(quintal, 0.1, 1.0))             # kg -> t, quintal (100 kg) -> t
value_t = num * scale * factor

df["loss_tonnes"]        = value_t.where(~per_ha & ~is_rate)
df["loss_tonnes_per_ha"] = (num * factor).where(per_ha & ~is_rate)
df["quantity_is_rate"]   = is_rate.fillna(False)                       # kg/quintal, kg/MT, per capita, heads
df["quantity_unit_assumed"] = q.notna() & ~unit_known                  # plain number -> assumed tonnes
df = df.drop(columns="loss_quantity")

world_food_loss = df
print("rates/counts (loss_tonnes left empty):", int(df["quantity_is_rate"].sum()))
print("plain numbers, tonnes assumed:", int(df["quantity_unit_assumed"].sum()))
display(df.loc[df["loss_quantity_text"].notna(), ["loss_quantity_text", "loss_tonnes", "loss_tonnes_per_ha", "quantity_is_rate"]].sample(10, random_state=1))
```

### **9. Standardize Food Supply Stages (fill from `activity`) and classify Measurement Scope**

```python
df = world_food_loss

print("missing stage before:", int(df["food_supply_stage"].isna().sum()))

stage_map = {
    "Farm": "Farm/Harvest", "Harvest": "Farm/Harvest",
    "Trader": "Wholesale/Market", "Wholesale": "Wholesale/Market", "Market": "Wholesale/Market",
    "Collector": "Wholesale/Market", "Distribution": "Wholesale/Market",
    "Packing": "Packing/Grading", "Grading": "Packing/Grading", "Stacking": "Packing/Grading",
}
df["food_supply_stage"] = df["food_supply_stage"].replace(stage_map)

KEYWORDS = [
    (r"stor",                                    "Storage"),
    (r"transport",                               "Transport"),
    (r"retail",                                  "Retail"),
    (r"consum|household",                        "Households"),
    (r"wholesale|market|trad|distribut|collect", "Wholesale/Market"),
    (r"process|mill|extract|grind|parboil",      "Processing"),
    (r"pack|grading",                            "Packing/Grading"),
    (r"export",                                  "Export"),
    (r"harvest|thresh|dry|shell|winnow|farm|field|reap|cut|clean", "Farm/Harvest"),
]
def activity_to_stage(text):
    if pd.isna(text):
        return pd.NA
    found = set()
    for token in re.split(r"[,;/&]|\band\b", str(text).lower()):
        for pat, stage in KEYWORDS:
            if re.search(pat, token):
                found.add(stage); break
    if len(found) == 1: return found.pop()
    if len(found) > 1:  return "Multiple stages"
    return pd.NA

miss = df["food_supply_stage"].isna()
df.loc[miss, "food_supply_stage"] = df.loc[miss, "activity"].map(activity_to_stage)
df["food_supply_stage"] = df["food_supply_stage"].fillna("Unknown stage")

df["measurement_scope"] = np.select(
    [df["food_supply_stage"].eq("Whole supply chain"),
     df["food_supply_stage"].eq("Post-harvest"),
     df["food_supply_stage"].eq("Multiple stages"),
     df["food_supply_stage"].eq("Unknown stage")],
    ["Whole chain", "Post-harvest (aggregate)", "Multi-stage", "Unknown"],
    default="Single stage"
)

world_food_loss = df
whole_chain_world_food_waste = world_food_loss[world_food_loss["measurement_scope"].eq("Whole chain")]
stage_world_food_waste       = world_food_loss[world_food_loss["measurement_scope"].eq("Single stage")]

print("missing stage after:", int((df["food_supply_stage"] == "Unknown stage").sum()))
display(df["food_supply_stage"].value_counts())
display(df["measurement_scope"].value_counts())
```

### **10. Commodity group from the CPC code**
> كود CPC مش كله بيبدأ بصفر: `01x` محاصيل، `02x` حيوانات ومنتجات خام، و`21` لـ `24` منتجات مصنّعة (لحوم، ألبان، طحين...).

```python
def cpc_group(code):
    c = str(code)
    crops = {"011": "Cereals", "012": "Vegetables", "013": "Fruits & nuts", "014": "Oilseeds",
             "015": "Roots & tubers", "016": "Beverage & spice crops", "017": "Pulses",
             "018": "Sugar crops", "019": "Other crops"}
    if c[:3] in crops:      return crops[c[:3]]
    if c.startswith("02"):  return "Animal products (raw)"
    if c.startswith("21"):  return "Meat, fish & oils (processed)"
    if c.startswith("22"):  return "Dairy & eggs (processed)"
    if c.startswith("23"):  return "Grain mill & other processed food"
    if c.startswith("24"):  return "Beverages & sugar (processed)"
    return "Other"

world_food_loss["commodity_group"] = world_food_loss["cpc_code"].map(cpc_group)
display(world_food_loss["commodity_group"].value_counts())
```

### **11. Flags: suspect year, zero loss, extreme values, literature outliers**
> **ليه flag ومش حذف؟** النسبة العالية مش بالضرورة غلط (فاقد الفاكهة في الحر ممكن يوصل 40%)، فبنعلّم ونسيب القرار في التحليل.

```python
df = world_food_loss

df["is_suspect_year"] = df["year"] > 2025
df["is_zero_loss"]    = df["loss_percentage"].eq(0)
df["loss_percentage_clean"] = df["loss_percentage"].mask(df["is_zero_loss"])
df["is_modelled"]     = df["method_data_collection"].str.contains("model", case=False, na=False)

# extreme value = above Q3 + 3*IQR inside its own commodity + scope (needs 20+ rows, countries only)
base = df.loc[~df["is_aggregate"]]
grp  = base.groupby(["commodity", "measurement_scope"])["loss_percentage"]
q3   = grp.transform(lambda x: x.quantile(0.75))
iqr  = grp.transform(lambda x: x.quantile(0.75) - x.quantile(0.25))
n    = grp.transform("size")
extreme = (n >= 20) & (base["loss_percentage"] > q3 + 3 * iqr)
df["is_extreme_value"] = extreme.reindex(df.index, fill_value=False)
df["is_lit_outlier"]   = df["is_extreme_value"] & df["method_data_collection"].eq("Literature Review")

world_food_loss = df
print(df[["is_suspect_year", "is_zero_loss", "is_modelled", "is_extreme_value", "is_lit_outlier"]].sum())
```

### **1. Detect copies and suspicious values, without deleting suspicious records**
> نفس فكرة مصر (New Valley وزيتون 2014)، بس معمّمة: صف محلي مطابق لصف قومي لنفس الدولة، وقيمة بعيدة 5 أضعاف عن وسيط نفس الدولة والمحصول في السلسلة الكاملة.

```python
df = world_food_loss

# --- subnational row identical to a national row of the same country (probably copied)
key = ["country", "commodity", "year", "food_supply_stage", "loss_percentage"]
national_idx = pd.MultiIndex.from_frame(df.loc[df["geo_level"].eq("National"), key])
df["is_subnational_copy"] = (df["geo_level"].eq("Subnational")
                             & pd.MultiIndex.from_frame(df[key]).isin(national_idx))

# --- whole-chain value 5x higher/lower than the median of the same country + commodity (3+ rows)
whole = df["food_supply_stage"].eq("Whole supply chain") & ~df["is_aggregate"]
g     = df.loc[whole].groupby(["country", "commodity"])["loss_percentage"]
med   = g.transform("median")
cnt   = g.transform("size")
ratio = df.loc[whole, "loss_percentage"] / med
flag  = (cnt >= 3) & (med > 0) & ((ratio < 0.2) | (ratio > 5))
df["is_suspect_value"] = flag.reindex(df.index, fill_value=False)

world_food_loss = df
print(df[["is_subnational_copy", "is_suspect_value"]].sum())
display(df.loc[df["is_subnational_copy"] | df["is_suspect_value"],
               ["country", "commodity", "year", "food_supply_stage", "subnational_region", "loss_percentage"]].head(20))
```

### **2. Handle missing values**
| Column | Why missing | Action |
|---|---|---|
| `treatment`, `cause_of_loss` | معظم الدراسات مسجلتهمش (94 لـ 97%) | fill with label **"Not specified"** |
| `subnational_region` | الصف قومي أو aggregate | fill with label **"Not applicable"** |
| `notes`, `activity`, `url`, `reference` | مفيش قيمة في المصدر | labels: **"No note" / "No activity" / "No url" / "No reference"** |
| `method_data_collection` | 36 صف | label **"Unknown"** |
| `loss_tonnes`, `loss_tonnes_per_ha`, `sample_size_n` | المصدر ماذكرهاش | **keep NaN** + indicator column |
| `loss_percentage_clean` | zero values turned into NaN | **keep NaN** |

Numeric columns are NOT imputed (mean/median would invent loss quantities that no source reported).

```python
df = world_food_loss

df["has_loss_tonnes"]        = df["loss_tonnes"].notna()
df["has_loss_tonnes_per_ha"] = df["loss_tonnes_per_ha"].notna()
df["has_sample_size"]        = df["sample_size_n"].notna()
df["has_cause_of_loss"]      = df["cause_of_loss"].notna()

fill_labels = {
    "treatment": "Not specified", "cause_of_loss": "Not specified", "sample_size_text": "Not specified",
    "notes": "No note", "activity": "No activity", "url": "No url", "reference": "No reference",
    "subnational_region": "Not applicable", "method_data_collection": "Unknown",
}
for col, label in fill_labels.items():
    df[col] = df[col].fillna(label)

# reliable = not a suspect year / copy / suspect value / zero / mismatch
# (literature + extreme values are NOT excluded: valid but lower-quality, so keep them flagged)
df["is_reliable"] = ~(df["is_suspect_year"] | df["is_subnational_copy"] | df["is_suspect_value"]
                      | df["is_zero_loss"] | df["is_pct_mismatch"])
df["is_analysis_ready"] = df["is_reliable"] & ~df["is_aggregate"]

world_food_loss = df
print("Reliable rows:", int(df["is_reliable"].sum()), "of", len(df))
print("Analysis-ready rows (reliable + not aggregate):", int(df["is_analysis_ready"].sum()))
display(df.isna().sum()[lambda s: s > 0])
```

### **3. Types, column order, validation**

```python
df = world_food_loss

cat_cols = ["country", "continent", "commodity", "commodity_group", "food_supply_stage",
            "measurement_scope", "geo_level", "subnational_region", "treatment",
            "method_data_collection", "reference"]
for c in cat_cols:
    df[c] = df[c].astype("category")

final_cols = [
    # what / when / where
    "country", "m49_code", "continent", "geo_level", "subnational_region",
    "cpc_code", "commodity", "commodity_group", "year",
    # stage
    "food_supply_stage", "measurement_scope", "activity", "treatment", "cause_of_loss",
    # values
    "loss_percentage", "loss_percentage_clean", "loss_percentage_original",
    "loss_tonnes", "has_loss_tonnes", "loss_tonnes_per_ha", "has_loss_tonnes_per_ha",
    "quantity_unit_assumed", "quantity_is_rate", "loss_quantity_text",
    "sample_size_n", "has_sample_size", "sample_size_text", "has_cause_of_loss",
    # quality flags
    "is_analysis_ready", "is_reliable", "is_aggregate", "is_suspect_year", "is_lit_outlier",
    "is_subnational_copy", "is_suspect_value", "is_zero_loss", "is_pct_mismatch",
    "is_range_value", "is_extreme_value", "is_modelled",
    # source
    "method_data_collection", "reference", "url", "notes",
]
df = df[final_cols]
for c in df.select_dtypes("boolean").columns:
    df[c] = df[c].fillna(False).astype(bool)
for c in df.select_dtypes("Float64").columns:
    df[c] = df[c].astype("float64")
n_dup = int(df.duplicated().sum())            # duplicates created by unifying country/commodity names
df = df.drop_duplicates().reset_index(drop=True)
print("duplicates removed after name unification:", n_dup)

# ---- validation (will raise an error if something is wrong) ----
assert df.duplicated().sum() == 0,                       "duplicate rows found"
assert df["loss_percentage"].between(0, 100).all(),      "loss_percentage out of 0-100"
cpc_ok = df["cpc_code"].str.fullmatch(r"\d{4,6}[A-Za-z]?(?:\.\d{1,2})?", na=False)   # 02351f, 021116i, 21111.01
if not cpc_ok.all():
    print("Unusual cpc_code values:"); print(df.loc[~cpc_ok, "cpc_code"].value_counts().head(20))
assert cpc_ok.all(),                                     "bad cpc_code format"
assert df["year"].between(2000, 2026).all(),             "year out of range"
key_cols = ["country", "cpc_code", "commodity", "year", "loss_percentage", "food_supply_stage",
            "measurement_scope", "geo_level", "method_data_collection", "reference"]
assert df[key_cols].notna().all().all(),                 "missing values in key columns"
assert df["country"].nunique() > 100,                    "countries were lost!"
print("All checks passed.", df.shape, "| countries:", df.loc[~df["is_aggregate"], "country"].nunique())

world_food_loss_final = df
display(df.head())
```

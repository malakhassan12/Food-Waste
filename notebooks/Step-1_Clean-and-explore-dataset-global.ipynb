## **Analysis Missing values in Data-set**

```python
missing_report = pd.DataFrame({
    "missing_count": food_global.isna().sum(),
    "missing_percentage": food_global.isna().mean().mul(100),
    "non_missing_count": food_global.notna().sum(),
    "unique_non_null": food_global.nunique(dropna=True)
})

missing_report = missing_report.sort_values(
    by="missing_percentage",
    ascending=False
)

display(missing_report)
```

### **Show "NA", "N/A", "NULL", "NONE", "UNKNOWN", "MISSING", "-" in string data-types**

```python
print("Empty strings:")
display(
    (food_global.astype("string").apply(
        lambda col: col.str.strip().eq("").sum()
    )).to_frame("empty_string_count")
)

missing_tokens = [
    "NA", "N/A", "NULL", "NONE", "UNKNOWN", "MISSING", "-"
]

for col in food_global.select_dtypes(include=["object", "string"]).columns:
    values = food_global[col].astype("string").str.strip().str.upper()
    count = values.isin(missing_tokens).sum()

    if count > 0:
        print(f"{col}: {count} possible missing tokens")
```

## **Describe numeric data**

```python
numeric_cols = food_global.select_dtypes(
    include=np.number
).columns

numeric_report = pd.DataFrame(index=numeric_cols)

numeric_report["missing"] = food_global[numeric_cols].isna().sum()
numeric_report["unique"] = food_global[numeric_cols].nunique()
numeric_report["min"] = food_global[numeric_cols].min()
numeric_report["max"] = food_global[numeric_cols].max()
numeric_report["mean"] = food_global[numeric_cols].mean()
numeric_report["median"] = food_global[numeric_cols].median()
numeric_report["std"] = food_global[numeric_cols].std()
numeric_report["negative_count"] = food_global[numeric_cols].lt(0).sum()

display(numeric_report)
```

## **Show ouliers**

```python
outlier_rows = []

for col in numeric_cols:
    series = food_global[col].dropna()

    if series.empty:
        continue

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    mask = (food_global[col] < lower_bound) | (
        food_global[col] > upper_bound
    )

    outlier_rows.append({
        "column": col,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": int(mask.sum()),
        "outlier_percentage": round(mask.mean() * 100, 2)
    })

outlier_report = pd.DataFrame(outlier_rows)
display(outlier_report)
```

## **Data quality check on the Year**

```python
if "Year" in food_global.columns:
    print("Invalid or suspicious years:")
    display(
        food_global.loc[
            ~food_global["Year"].between(1900, 2100),
            ["Year"]
        ].head(20)
    )
```

## **Finally analysis of data**

```python
quality_report = pd.DataFrame(index=food_global.columns)

quality_report["dtype"] = food_global.dtypes.astype(str)
quality_report["total_rows"] = len(food_global)
quality_report["missing_count"] = food_global.isna().sum()
quality_report["missing_pct"] = (
    food_global.isna().mean() * 100
).round(2)

quality_report["unique_count"] = food_global.nunique(dropna=True)
quality_report["unique_pct"] = (
    food_global.nunique(dropna=True) / max(len(food_global), 1) * 100
).round(2)

quality_report["duplicate_row_count"] = food_global.duplicated().sum()
quality_report["memory_bytes"] = food_global.memory_usage(deep=True).iloc[1:].values

display(quality_report)
```
